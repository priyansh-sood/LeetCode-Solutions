```python

faqs = """About Artificial Intelligence

What is Artificial Intelligence?
Artificial Intelligence is a branch of computer science that enables machines to perform tasks that normally require human intelligence.

What is Machine Learning?
Machine Learning is a subset of Artificial Intelligence that allows computers to learn patterns from data and make predictions.

What is Deep Learning?
Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers to learn complex patterns.

What is Natural Language Processing?
Natural Language Processing is a field of Artificial Intelligence that helps computers understand and generate human language.

What is an LSTM?
LSTM stands for Long Short-Term Memory. It is a type of recurrent neural network designed to learn patterns in sequential data.

What is next-word prediction?
Next-word prediction is a language modeling task in which a model predicts the most likely word to follow a given sequence of words.

How does an LSTM predict the next word?
An LSTM processes the input sequence, learns contextual patterns, and predicts a probability distribution over possible next words.

What is tokenization?
Tokenization is the process of splitting text into smaller units called tokens, such as words or subwords.

What is a vocabulary in NLP?
A vocabulary is the collection of unique tokens that a language model can recognize.

What is an embedding layer?
An embedding layer converts token IDs into dense numerical vectors that a neural network can learn from.

What is a neural network?
A neural network is a machine learning model made of interconnected units that learn patterns by adjusting their weights.

What is an epoch?
An epoch is one complete pass through the entire training dataset.

What is overfitting?
Overfitting occurs when a model learns the training data too closely and performs poorly on unseen examples.

What is a dataset?
A dataset is a collection of data used to train, validate, or test a machine learning model.
"""
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([faqs])
```

```python
tokenizer.word_index
```

```python
for sentence in faqs.split('\n'):
  print(sentence)
```

```python
input_sequences =[]
for sentence in faqs.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]
  for i in range(1, len(tokenized_sentence)):
    input_sequences.append ( tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len = max ([len (x)for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences (input_sequences, maxlen = max_len,padding = 'pre')
```

```python
padded_input_sequences
```

```python
x = padded_input_sequences [:,:-1]
```

```python
y = padded_input_sequences [:,-1]
```
